# Artificial Neural Networks 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import torch
import torch.nn as nn           #ngambil neural network, ngambil loss function
import torch.optim as optim     # Ngambil optimizer

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
```

```python
iris = load_iris()
X = iris.data
y = iris.target
```

```python
scaler = MinMaxScaler()
x = scaler.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
X_train = torch.tensor(X_train, dtype=torch.float32)
y_train = torch.tensor(y_train, dtype=torch.long)       #long int -> bentuk int lebih besar
X_test = torch.tensor(X_test, dtype=torch.float32)
y_test = torch.tensor(y_test, dtype=torch.long)
```

```python
# # dipelajari ulang dan tonton video nya

# model = nn.Sequential(
#     nn.Linear(4, 32),     # x / input layer kita ada 4 node, kita coba hidden layer 32
#     nn.ReLU(),
#     nn.Linear(32, 16),    # kita mau berapa hidden layer
#     nn.ReLU(),
#     nn.Linear(16, 3),     # hidden layer ke-2 ada 16 node, output layer ada 3
#     # nn.Softmax()
# )
# # input node 4 hidden layer pertama 4 -> 32, hidden layer kedua 32 -> 16, output ketiga 16 -> 3
# # tonton ulang dimenit 9.26
```

```python
model = nn.Sequential(
    nn.Linear(4, 8),
    nn.ReLU(),
    nn.Linear(8, 3),
    # nn.Softmax() 
)
```

```python
loss_fun = nn.CrossEntropyLoss()    # sudah include Softmax()
optimizer = optim.SGD(model.parameters(), lr=0.1)    # Stochastic Gradient Descent, ini dipake pas zoom

# optimizer = optim.Adam(model.parameters(), lr=0.1)
```

ada 3 tahapan di neural network
1. Feed forward
2. Compute loss
3. back propagation

diawal kita memasukan semua data kita sampe mendapatkan output, setelah outputnya dapat kita compute loss nya.kita akan menghitung seberapa salah model tersebut di compute loss. setalah tau loss kita gunakan compute loss itu pada posisi back propagation.<br>
fungsi dari back propagation memberitahu semua pertama kali maju sampai mendapatkan sebuah output, setelah dapat output kembali ke belakang fungsi nya untuk menjadi sumber evaluasi, kira-kira berdasarkan loss tersebut dibagian mana yg memberikan dampak loss tertinggi.<br>

dan semua ini akan diulang berkali-kali sebanyak epoch

```python
epochs = 50                 # model setiap epoch itu akan belajar. setiap kali belajar itu harus di reset gradient nya
for epoch in range(epochs):     # karena kita ga ingin pembelajaran sebelumnya itu mempengaruhi pembelajaran saat ini. nyatanya emang mempengaruhi tapi dia tidak mengikut campuri pengaruh tersebut
    optimizer.zero_grad()       # gradient descent ke reset
    ouputs = model(X_train)
    loss = loss_fun(ouputs, y_train)
    loss.backward()
    optimizer.step()            # menjalankan gradient descent

    with torch.no_grad():
        test_output = model(X_test)
        _, predicted = torch.max(test_output, 1)    # _ diabaikan
        acc = (predicted == y_test).sum() / len(y_test)
    
    print(f'Epoch - {epoch + 1}/{epochs}. Loss: {loss.item():.4f}. Test Acc: {acc*100:.4f}%')
```
