```python
import sys, subprocess, importlib, os
```

```python
def pip_install(*packages):
    """Install/upgrade packages via pip using the current Python executable."""
    print("Installing:", " ".join(packages))
    subprocess.run([sys.executable, "-m", "pip", "install", "--upgrade", *packages], check=True)
```

```python
def try_import(name):
    try:
        return importlib.import_module(name)
    except Exception as e:
        print(f"[WARN] Import {name} gagal: {e}")
        return None
```

```python
# 1) Install paket inti
pip_install("pip", "ultralytics", "opencv-python", "ffmpeg-python")
```

```python
# 2) Import & cek versi
torch = try_import("torch")            # Colab biasanya sudah punya torch
cv2   = try_import("cv2")
```

```python
if torch is None:
    print("Memasang PyTorch CPU-only (fallback)...")
    pip_install("torch", "torchvision", "--index-url", "https://download.pytorch.org/whl/cpu")
    torch = importlib.import_module("torch")
```

```python
YOLO = None
try:
    from ultralytics import YOLO as _YOLO
    YOLO = _YOLO
    print("Ultralytics terimport ✅")
except Exception as e:
    print("[WARN] Ultralytics import gagal:", e)
```

```python
cuda_ok = getattr(torch, "cuda", None) and torch.cuda.is_available()
print("Versions ->")
print("  torch     :", getattr(torch, "__version__", None))
print("  CUDA avail:", bool(cuda_ok))
print("  opencv    :", getattr(cv2, "__version__", None))
```
