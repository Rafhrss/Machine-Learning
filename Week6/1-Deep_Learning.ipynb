# Deep Learning / Neural Network

minimal ada 3 di deep learning
1. jumlah node yg ada di input layer memiliki jumlah yg sama dengan feature kita, lalu dibawa ke dimensi yg lebih gede
2. hidden layer
3. output layer, jumlah output layer sama dengan jumlah class kita yg kita ingin kan

Bisa membuat lebih dari 3 layer diatas seperti menambah 2 hidden layer lagi, tapi ada resiko nya

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def f_1(x):
    return x ** 2 + 2 * x + 1
```

```python
def f_prime_1(x):
    return 2 * x + 2
```

```python
x = 5.0
lr = 0.1
```

```python
X_history = []
y_history = []
grad_history = []
```

```python
for step in range(20):
    y = f_1(x)
    grad = f_prime_1(x)
    x = x - lr * grad   #gradient descent formula
    
    X_history.append(x)
    y_history.append(y)
```

```python
X = np.linspace(-5, 5, 100)
Y = f_1(X)
y_prime = f_prime_1(X)

plt.figure(figsize=(8,5))
plt.plot(X, Y, label="f(x) = x^2 + 2x + 1")
plt.plot(X, y_prime, label="f(x) = 2x + 2")
plt.scatter(X_history, y_history, color="red", label="Gradient Descent Steps")
plt.plot(X_history, y_history, linestyle="--", color="red")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Gradient Descent on f(x) = x^2 + 2x + 1 (NumPy Version)")
plt.legend()
plt.grid(True)
plt.show()
```

## Experiment

```python
# x^2 + 2x + 1
def f(x):
    return 4*x**3 - 9*x**2
```

```python
# 2x + 2
def f_prime(x):
    return 12*x**2 - 18*x
```

```python
# Visualisasi fungsi
x_dist = np.linspace(-2, 3, 500)
y = f(x_dist)
y_prime = f_prime(x_dist)

plt.plot(x_dist, y, label="f(x)")
plt.plot(x_dist, y_prime, label="f'(x)")
plt.axhline(0, color="black", linewidth=0.5)
plt.title("Contoh Fungsi dengan Local & Global Minima")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
plt.show()
```

```python
x_gatau = np.arange(-2, 3, 0.1)
```

```python
x = np.random.choice(x_gatau)
lr = 0.01
```

```python
x
```

```python
x_history = []
y_history = []
grad_history = []
```

```python
for step in range(20):
    y = f(x)
    grad = f_prime(x)
    x = x - lr * grad           #gradient descent formula

    x_history.append(x)
    y_history.append(y)
```

```python
x_dist = np.linspace(-1, 3, 500)
y = f(x_dist)
y_prime = f_prime(x_dist)

plt.figure(figsize=(8,5))
plt.plot(x_dist, y, label="f(x) = x^2 + 2x + 1")
plt.plot(x_dist, y_prime, label="f(x) = 2x + 2")
plt.scatter(x_history, y_history, color="red", label="Gradient Descent Steps")
plt.plot(x_history, y_history, linestyle="--", color="red")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Gradient Descent on f(x) = x^2 + 2x + 1 (NumPy Version)")
plt.legend()
plt.grid(True)
plt.show()
```
